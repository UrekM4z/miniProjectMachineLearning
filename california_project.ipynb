```python
import numpy as np
# Remarque : on utilise fetch_ au lieu de load_ car le dataset est un peu plus lourd
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

# 1. Chargement des données
print("Téléchargement/Chargement des données en cours...")
california = fetch_california_housing()
X = california.data
y = california.target

# 2. Découpage Train/Test (80/20)
# On garde le fameux random_state=42 pour la reproductibilité
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)

# 3. Baseline : Entraînement de la Régression Linéaire classique
modele_baseline = LinearRegression()
modele_baseline.fit(X_train, y_train)

# 4. Prédiction et Évaluation sur le Test
pred_test = modele_baseline.predict(X_test)
rmse_test = np.sqrt(mean_squared_error(y_test, pred_test))
r2_test = r2_score(y_test, pred_test)

print("\n--- Baseline : Régression Linéaire ---")
print(f"RMSE Test : {rmse_test:.4f}")
print(f"R² Test   : {r2_test:.4f}")
```

```python
from sklearn.tree import DecisionTreeRegressor

# 1. L'arbre non bridé (il va pousser jusqu'à apprendre par cœur)
arbre_fou = DecisionTreeRegressor(random_state=42)
arbre_fou.fit(X_train, y_train)

rmse_train_fou = np.sqrt(mean_squared_error(y_train, arbre_fou.predict(X_train)))
rmse_test_fou = np.sqrt(mean_squared_error(y_test, arbre_fou.predict(X_test)))

# 2. L'arbre bridé (on le limite à 5 niveaux de profondeur)
arbre_sage = DecisionTreeRegressor(max_depth=5, random_state=42)
arbre_sage.fit(X_train, y_train)

rmse_train_sage = np.sqrt(mean_squared_error(y_train, arbre_sage.predict(X_train)))
rmse_test_sage = np.sqrt(mean_squared_error(y_test, arbre_sage.predict(X_test)))

print("--- Arbre de Décision (Sans limite) ---")
print(f"RMSE Train : {rmse_train_fou:.4f}")
print(f"RMSE Test  : {rmse_test_fou:.4f}\n")

print("--- Arbre de Décision (max_depth=5) ---")
print(f"RMSE Train : {rmse_train_sage:.4f}")
print(f"RMSE Test  : {rmse_test_sage:.4f}")
```

```python
from sklearn.tree import DecisionTreeRegressor

# 1. L'arbre non bridé (il va pousser jusqu'à apprendre par cœur)
arbre_fou = DecisionTreeRegressor(random_state=42)
arbre_fou.fit(X_train, y_train)

rmse_train_fou = np.sqrt(mean_squared_error(y_train, arbre_fou.predict(X_train)))
rmse_test_fou = np.sqrt(mean_squared_error(y_test, arbre_fou.predict(X_test)))

# 2. L'arbre bridé (on le limite à 5 niveaux de profondeur)
arbre_sage = DecisionTreeRegressor(max_depth=5, random_state=42)
arbre_sage.fit(X_train, y_train)

rmse_train_sage = np.sqrt(mean_squared_error(y_train, arbre_sage.predict(X_train)))
rmse_test_sage = np.sqrt(mean_squared_error(y_test, arbre_sage.predict(X_test)))

print("--- Arbre de Décision (Sans limite) ---")
print(f"RMSE Train : {rmse_train_fou:.4f}")
print(f"RMSE Test  : {rmse_test_fou:.4f}\n")

print("--- Arbre de Décision (max_depth=5) ---")
print(f"RMSE Train : {rmse_train_sage:.4f}")
print(f"RMSE Test  : {rmse_test_sage:.4f}")
```

```python
from sklearn.ensemble import RandomForestRegressor

# 1. On initialise la forêt avec 100 arbres
foret = RandomForestRegressor(n_estimators=100, random_state=42)

# 2. Entraînement (ça peut prendre 2 ou 3 secondes de plus, il y a 100 arbres à faire pousser !)
foret.fit(X_train, y_train)

# 3. Prédictions sur le jeu de test
pred_test_foret = foret.predict(X_test)

# 4. Calcul des métriques
rmse_test_foret = np.sqrt(mean_squared_error(y_test, pred_test_foret))
r2_test_foret = r2_score(y_test, pred_test_foret)

print("--- Forêt Aléatoire (100 arbres) ---")
print(f"RMSE Test : {rmse_test_foret:.4f}")
print(f"R² Test   : {r2_test_foret:.4f}")
```

```python
from sklearn.ensemble import GradientBoostingRegressor

# 1. On initialise le modèle (par défaut, il va créer 100 arbres séquentiels)
boosting = GradientBoostingRegressor(random_state=42)

# 2. Entraînement (il apprend de ses erreurs, étape par étape)
boosting.fit(X_train, y_train)

# 3. Prédictions sur le jeu de test
pred_test_boost = boosting.predict(X_test)

# 4. Calcul des métriques
rmse_test_boost = np.sqrt(mean_squared_error(y_test, pred_test_boost))
r2_test_boost = r2_score(y_test, pred_test_boost)

print("--- Gradient Boosting ---")
print(f"RMSE Test : {rmse_test_boost:.4f}")
print(f"R² Test   : {r2_test_boost:.4f}")
```

```python
import pandas as pd

# 1. On récupère les noms des caractéristiques (les colonnes)
noms_features = california.feature_names

# 2. On extrait le score d'importance calculé par la Forêt Aléatoire
importances = foret.feature_importances_

# 3. On met tout ça au propre dans un tableau Pandas, qu'on trie du plus grand au plus petit
df_importances = pd.DataFrame({
    'Variable': noms_features,
    'Importance': importances
}).sort_values(by='Importance', ascending=False)

# On affiche les pourcentages d'importance
print("--- Importance des Variables (Random Forest) ---")
df_importances['Importance'] = df_importances['Importance'].apply(lambda x: f"{x*100:.1f}%")
print(df_importances.to_string(index=False))
```
